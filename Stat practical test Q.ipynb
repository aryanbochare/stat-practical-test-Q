<h2>Ex.1.Check the goodness of fit for the following data mapping number of hours a student studies vs the expected number of hours as 5% los.
<h2>obseverd data (in hr) 8,6,10,7,8,11,9
<h2>Expected data(in hr ) 9,8,11,8,10,7,6

```python

from scipy.stats import chisquare

observed = [8, 6, 10, 7, 8, 11, 9]

expected = [9, 8, 11, 8, 10, 7, 6]

chi, p = chisquare(f_obs=observed, f_exp=expected)

print("Chi-square value =", chi)
print("P-value =", p)

if p > 0.05:
    print("Good fit (Accept H0)")
else:
    print("Not a good fit (Reject H0)")
    
```

<h1>Ex.2 What is the probability of 2 heads in 6 coin tosses? fit the Binomial distribution for the coin tosses and plot it

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

x = range(7)
y = binom.pmf(x, 6, 0.5)

print("Probability of 2 heads =", binom.pmf(2, 6, 0.5))

plt.bar(x, y)
plt.show()
```
